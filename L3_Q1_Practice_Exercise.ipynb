### Join (Merge)

```python
left = pd.DataFrame({"key": ["foo", "foo"], "lval": [1, 2]})
right = pd.DataFrame({"key": ["foo", "foo"], "rval": [4, 5]})
pd.merge(left, right, on="key") # inner join

# [NOTE]: for same key values, the merge function will create a Cartesian product of the rows.
```

```python
left = pd.DataFrame({"key": ["foo", "bar"], "lval": [1, 2]})
right = pd.DataFrame({"key": ["foo", "bar"], "rval": [4, 5]})
pd.merge(left, right, on="key")
```

### Reshaping (stack, unstack)

```python
arrays = [
   ["bar", "bar", "baz", "baz", "foo", "foo", "qux", "qux"],
   ["one", "two", "one", "two", "one", "two", "one", "two"],
]
tuples = list(zip(*arrays))

index = pd.MultiIndex.from_arrays(arrays, names=["first", "second"])
# index = pd.MultiIndex.from_tuples(tuples, names=["first", "second"])

df = pd.DataFrame(np.random.randn(8, 2), index=index, columns=["A", "B"])
df
```

```python
stacked = df.stack() # Compress DataFrame columns into rows
stacked
```

```python
stacked.unstack() # Add a new level to the index, unstacking the last level
```

### Categorical

```python
df = pd.DataFrame({
    "id": [1, 2, 3, 4, 5, 6],
    "raw_grade": ["a", "b", "b", "a", "a", "e"]
})
df.info()
```

```python
# Add 'raw grades' as categorical datatype in new column
df["grade"] = df["raw_grade"].astype("category")
df["grade"]
```

```python
new_categories = ["very good", "good", "very bad"]
df["grade"] = df["grade"].cat.rename_categories(new_categories)
df["grade"]
```

```python
# Reorder the categories and simultaneously add the missing categories
df["grade"] = df["grade"].cat.set_categories(
    ["very bad", "bad", "medium", "good", "very good"]
)
df["grade"]
```

```python
df.sort_values(by="grade")
```

```python
df.groupby("grade", observed=False).size()

# [NOTE]: Grouping by a categorical column with 'observed=False' also shows empty categories
```
