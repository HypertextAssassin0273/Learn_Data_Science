##### 1) Load the data

```python
df_book = pd.read_csv("data/bl_flickr_images_book.csv")
df_book.head()
```

##### 2) Dropping unnecessary columns

We can see some columns provide ancillary information that would be useful
for the library but aren't very descriptive of the books.
Drop those columns to have just the information about the books.

```python
# list of columns to be dropped
to_drop = [
    'Edition Statement', 'Corporate Author', 'Corporate Contributors',
    'Former owner', 'Engraver', 'Contributors', 'Issuance type',
    'Shelfmarks'
]

df_book.drop(columns=to_drop, inplace=True)
df_book.head()
```

##### 3) Set 'Identifier' column as index

First of all, check if the column is unique or not.

```python
df_book['Identifier'].is_unique
```

In order to make sure it would be a good idea to use this column as index,
let's see at some example.

```python
df_book.loc[0]
```

It can be noticed that the 1st record in our dataframe has the identifier **206** which can be accessed in a straightforward way with `loc[]`.

`loc[]` allows us to to label-based indexing (_labeling of a row regardless of
its position_), so why not use the 'Identifier' column as index so it's easier

##### 4) Cleaning fields in the data (dtypes and data inconsistencies)

It's important to get specific columns like date to a uniform format for
better understanding and enforce consistency.

Let's first check the different datatypes we have in our dataset.

```python
df_book.dtypes.value_counts() # [NOTE]: DataFrame.get_dtype_counts() is deprecated!
```

```python
df_book.dtypes
```

You can see that all of the datatypes are currently **'object'** in native Python.

**'Date of Publication'** is one column where it makes sense to enforce a **numeric
value** so calculations can be performed later on.

```python
# let's check a few values first
df_book['Date of Publication'].head(15)
```

We can notice a few issues here:
1. **extra dates in []:** 1879 [1868]
2. **date ranges:** 1839, 38-54
3. **NaN**s

We can fix these with **regex** as we notice that we only need the **first 4 digits** to get our year correct.

```python
regex = r'^(\d{4})' # to find any 4 digits at the beginning
date_num = df_book['Date of Publication'].str.extract(regex, expand=False)
date_num.head(15)
```

Now that we've got our desired year values, the **dtype** of this column is still
**'object'**. let's make it **numeric**.

```python
df_book['Date of Publication'] = pd.to_numeric(date_num, downcast='integer')
df_book['Date of Publication'].dtype
```

We still have **NaN**s, let's check how many we have:

```python
no_of_null_values = int(df_book['Date of Publication'].isnull().sum())
no_of_null_values # 971
```

Maybe a **percentage** over the whole set is better to decide if it's problematic:

```python
no_of_null_values / len(df_book) * 100
```

**~11.72%** of our data is **missing** which is not a big problem for us right now.
but still we can just skip those records:

```python
df_book = df_book[df_book['Date of Publication'].isnull() != True]
df_book.head(15)
```

Now let's check a (_possible_) **string** column **'Place of Publication'** for inconsistencies:

```python
df_book['Place of Publication'].head(15)
```

We see that for some rows, the **'place of publications'** contains some unnecessary
information. further check leads us to identify **'London'** and **'Oxford'** to be
the values that we are actually concerned with.

```python
# comparing 2 entries for another issue
book_indexes = [4157862, 4159587]
df_book.loc[book_indexes]
```

Here, we see that both values are the same but one of them is separated using **'-'**.

Let's fix all the records for the aforementioned cases with **str** methods combined with **numpy**:

###### _toc reference_

```python
# combining 'str' methods with numpy to clean columns
df_book['Place of Publication'] = np.where(
    df_book['Place of Publication'].str.contains('London'),
    'London',
    df_book['Place of Publication'].str.replace('-', ' ')
)
```

Let's test specific examples that we know earlier were to be fixed:

```python
book_indexes = [216, 4157862, 4159587]
df_book.loc[book_indexes]
```

let's check the values again:

```python
df_book['Place of Publication'].head(15)
```

And that's all for this dataset.

### **Part 'B':** 'University_Towns.txt' Dataset
- [Cleaning the Entire Dataset Using the applymap Function](######toc-reference-2)

Let’s look at an example. We will create a DataFrame out of the **'university_towns.txt'** file:

```python
%%bash
head data/university_towns.txt
```

Taking advantage of the structure of the file to extract the data into a list of **'(state, city)'** tuples:

```python
university_towns = []

with open('data/university_towns.txt') as file:
    for line in file:
        if '[edit]' in line:
            # Remember this `state` until the next is found
            state = line
        else:
            # Otherwise, we have a city; keep `state` as last-seen
            university_towns.append((state, line))

university_towns[:5]
```

Now we'll convert the list into a DataFrame:

```python
df_towns = pd.DataFrame(university_towns, columns=['State', 'RegionName'])
df_towns.head()
```

Instead of cleaning these strings in **for loop** with `.str` methods, we can use the **`.map`** method to apply a function element-wise in a DataFrame.

###### _toc reference 2_

```python
def get_citystate(item):
    if ' (' in item:
        return item[:item.find(' (')]
    elif '[' in item:
        return item[:item.find('[')]
    else:
        return item

df_towns = df_towns.map(get_citystate) # [NOTE]: DataFrame.applymap() is deprecated!
df_towns.head()
```

**Note:** while it's a convenient and versatile method, `.map` can have significant runtime for larger datasets, because it maps a Python callable to each individual element. In some cases, it can be more efficient to do _`vectorized`_ operations that utilize _`Cython`_ or _`NumPY`_ (which, in turn, makes calls in _`C`_ language) under the hood.
