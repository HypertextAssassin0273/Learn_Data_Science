### Implementing PCY Algorithm on Online Retail Dataset

```python
import pandas as pd
from collections import defaultdict
import itertools
```

```python
data = pd.read_excel("data/online_retail.xlsx")
```

```python
data_clean = data.dropna(subset=['InvoiceNo', 'StockCode'])
```

```python
transactions = data_clean.groupby('InvoiceNo')['StockCode'].apply(list).tolist()
```

```python
item_counts = defaultdict(int)
for transaction in transactions:
    for item in transaction:
        item_counts[item] += 1
```

```python
min_support = 2
frequent_items = {item for item, count in item_counts.items() if count >= min_support}
```

```python
hash_table = defaultdict(list)
pair_counts = defaultdict(int)
```

```python
bucket_size = 100
def hash_pair(pair):
    return hash(pair) % bucket_size
```

```python
for transaction in transactions:
    transaction = [item for item in transaction if item in frequent_items]

    pairs = itertools.combinations(transaction, 2)
    for pair in pairs:
        bucket = hash_pair(pair)
        hash_table[bucket].append(pair)
        pair_counts[pair] += 1
```

```python
bucket_info = []

for bucket, pairs in hash_table.items():
    pair_support = defaultdict(int)

    for pair in pairs:
        pair_support[pair] += 1

    highest_support = max(pair_support.values(), default=0)

    highest_support_pairs = [pair for pair, support in pair_support.items() if support == highest_support]

    for pair in highest_support_pairs:
        bucket_info.append({
            "Bit Vector": 1,
            "Bucket No.": bucket,
            "Highest Support Count": highest_support,
            "Pairs": pair,
            "Candidate Set": pair
        })
```

```python
bucket_df = pd.DataFrame(bucket_info)
bucket_df
```
