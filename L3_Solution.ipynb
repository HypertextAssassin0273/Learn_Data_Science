```python
# Import libraries
import numpy as np
import pandas as pd
from collections import Counter
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

### **Question: 1**

```python
# Part (a)
names = ['sepal-length', 'sepal-width', 'petal-length', 'petal-width', 'class']
training = pd.read_csv("data/iris.csv", names=names, header=0)
training.head()
```

```python
# Part (b)
print('Size:', training.shape)
```

```python
# Part (c)
print('class distribution\n', training.groupby('class').size())
```

```python
# Part (d)
array = training.values
X = array[:,0:4]
Y = array[:,4]
t_size = 0.20
seed = 7

X_train_I, X_test_I, Y_train_I, Y_test_I = train_test_split(X, Y, test_size=t_size, random_state=seed)
```

```python
# Part (e)
knn = KNeighborsClassifier()
knn.fit(X_train_I, Y_train_I)
predictions = knn.predict(X_test_I)
print('score: ', accuracy_score(Y_test_I, predictions))
print('\nconfusion_matrix:\n', confusion_matrix(Y_test_I, predictions))
print('\nclassification_report:\n', classification_report(Y_test_I, predictions))
```

```python
# Part (f)
for i in range(1, 11):
    f = KNeighborsClassifier(n_neighbors=i)
    f.fit(X_train_I, Y_train_I)
    f_pred = f.predict(X_test_I)
    print(f"For {i} neighbors: {accuracy_score(Y_test_I, f_pred):.3f}")
```

```python
# Part (g)
for i in range(1, 11):
    X_train_I, X_test_I, Y_train_I, Y_test_I = train_test_split(X, Y, test_size=t_size, random_state=i)

    g = KNeighborsClassifier()
    g.fit(X_train_I, Y_train_I)
    f_pred = g.predict(X_test_I)
    print(f"For seed {i}: {accuracy_score(Y_test_I, f_pred):.3f}")
```

### **Question: 3**

```python
# Calculate (Vectorized) Chi-squared distance
def get_chi_squared_distance(X_test, X_train):
    X_test = np.asarray(X_test, dtype=float)
    X_train = np.asarray(X_train, dtype=float)
    # Broadcasting: (n_test, n_train, n_features)
    diff = X_test[:, np.newaxis, :] - X_train[np.newaxis, :, :]
    sum_ab = X_test[:, np.newaxis, :] + X_train[np.newaxis, :, :] + 1e-10
    return np.sum((diff ** 2) / sum_ab, axis=2)

# Optimized Custom kNN classifier
def custom_kNN(X_train, Y_train, X_test, k):
    distances = get_chi_squared_distance(X_test, X_train) # Shape: (n_test, n_train)
    indices = np.argsort(distances, axis=1)[:, :k] # Shape: (n_test, k)
    k_classes = Y_train[indices] # Shape: (n_test, k)
    predictions = [Counter(row).most_common(1)[0][0] for row in k_classes] # count occurrences to get the most common class
    return np.array(predictions)
```
